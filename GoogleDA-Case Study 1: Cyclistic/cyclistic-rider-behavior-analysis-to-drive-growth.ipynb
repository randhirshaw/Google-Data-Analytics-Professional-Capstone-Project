# 🧰 Importing Necessary Libraries

```python
# Core data manipulation and numerical operations
import pandas as pd
import numpy as np

# File and directory handling (used for loading multiple monthly datasets)
import os

# Visualization libraries (used in Analyze and Share phases)
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings for improved readability during exploration
pd.set_option('display.max_columns', None)
pd.set_option('display.float_format', '{:.2f}'.format)

# Set a consistent visualization style for clarity and comparability
sns.set_theme(style="whitegrid")

# Suppress non-critical future warnings to keep output clean
# (Other warnings are left visible to avoid masking potential issues)
import warnings
warnings.filterwarnings(
    "ignore",
    category=FutureWarning
)
```

# 📥 Load and Combine Monthly Cyclistic Trip Data

```python
# Root directory containing monthly CSV files
# (Update this path as needed for local, Kaggle, or cloud environments)
root_dir = "/kaggle/input/cyclistic-case-study-dataset"

# Identify all CSV files in the directory
csv_files = sorted(
    os.path.join(root_dir, file)
    for file in os.listdir(root_dir)
    if file.lower().endswith(".csv")
)

# Safety check to ensure data files are available
if not csv_files:
    raise FileNotFoundError(f"No CSV files found in directory: {root_dir}")

# Load each monthly dataset and track schema
dataframes = []
column_sets = []

for file in csv_files:
    df = pd.read_csv(file, low_memory=False)
    df["source_file"] = os.path.basename(file)  # Enable traceability
    dataframes.append(df)
    column_sets.append(set(df.columns))

# Validate schema consistency across files
unique_schemas = set(map(frozenset, column_sets))
if len(unique_schemas) > 1:
    print("⚠️ Warning: Column mismatches detected across monthly files.")
    print("Schemas will be standardized during the Process phase.")
else:
    print("✅ Column schema is consistent across all monthly files.")

# Combine all monthly datasets into a single DataFrame
rides_df = pd.concat(dataframes, ignore_index=True)

# Initial sanity checks
print(f"Number of CSV files merged: {len(csv_files)}")
print(f"Final dataset shape: {rides_df.shape}")

# Preview combined dataset
rides_df.head()
```

### Missing Data Handling

Exploratory review revealed missing values primarily in station-related fields:

This approach ensures that missing station data does not bias rider behavior comparisons while maintaining transparency around analytical scope and limitations.

### Data Type Standardization and Duplicate Removal

To ensure consistency and support reliable time-based analysis, key data types were reviewed and standardized prior to feature engineering.

The `started_at` and `ended_at` columns were converted to datetime format to enable accurate ride duration calculations, day-of-week extraction, and hourly usage analysis. During conversion, mixed timestamp formats were identified—some records included millisecond precision while others did not. Datetime parsing was therefore performed using Pandas’ mixed-format handling to ensure all valid timestamps were correctly converted without data loss.

Station identifier fields (`start_station_id`, `end_station_id`) were standardized as string values to preserve their categorical nature and prevent unintended numeric operations.

Duplicate ride records were identified based on complete row duplication and removed to avoid inflated ride counts and distorted summary statistics. The impact of duplicate removal was verified to confirm that only redundant records were excluded and that no valid trips were affected.

These steps establish a clean, consistent dataset suitable for downstream behavioral analysis.


```python
# 🔧 Fix Data Types and Remove Duplicates

# Parse datetime columns with support for mixed formats
rides_df["started_at"] = pd.to_datetime(
    rides_df["started_at"],
    format="mixed",
    errors="coerce"
)

rides_df["ended_at"] = pd.to_datetime(
    rides_df["ended_at"],
    format="mixed",
    errors="coerce"
)

# Remove records with invalid or missing timestamps
before_timestamps = len(rides_df)
rides_df = rides_df.dropna(subset=["started_at", "ended_at"])
print("Rows removed due to invalid timestamps:", before_timestamps - len(rides_df))

# Remove fully duplicated records
before_duplicates = len(rides_df)
rides_df = rides_df.drop_duplicates()
print("Duplicate rows removed:", before_duplicates - len(rides_df))

# Standardize data types
rides_df["start_station_id"] = rides_df["start_station_id"].astype("string")
rides_df["end_station_id"] = rides_df["end_station_id"].astype("string")
rides_df["member_casual"] = rides_df["member_casual"].astype("category")

# Verify dtype changes
rides_df[["start_station_id", "end_station_id", "member_casual"]].dtypes
```

Categorical fields were standardized to appropriate data types. Station identifier columns were converted to Pandas’ nullable string type to preserve missing values, and rider classification was converted to a categorical type to support efficient grouping and comparison.

### Feature Engineering

To support comparison of riding behavior between casual riders and annual members, a focused set of time- and duration-based features was engineered from trip timestamps. These features directly enable analysis of **when**, **how long**, and **how frequently** riders use Cyclistic bikes.

The following features were created:

- **ride_length_min** — Ride duration in minutes, calculated as the difference between ride end and start times  
- **day_of_week** — Day of the week when the ride started, used to compare weekday and weekend behavior  
- **hour** — Hour of the day when the ride started, used to identify daily usage patterns  
- **month** — Month when the ride started, used to capture seasonal trends  

Records with invalid ride durations (zero or negative values) are excluded prior to analysis to prevent distortion of aggregate metrics. No additional spatial features or categorical encodings were introduced, as location-based analysis is outside the scope of this behavioral study.

These engineered features form the foundation for the comparative analysis conducted in the **Analyze phase**.


```python
def create_features(df):
    """
    Create time-based and duration-based features required for behavioral analysis.
    Assumes datetime fields have already been validated and cleaned.
    """
    df = df.copy()

    # Calculate ride duration in minutes
    df["ride_length_min"] = (
        (df["ended_at"] - df["started_at"])
        .dt.total_seconds()
        .div(60)
    )

    # Remove invalid rides (zero or negative duration)
    before_duration_filter = len(df)
    df = df[df["ride_length_min"] > 0]
    print(
        f"Removed {before_duration_filter - len(df)} rides with non-positive duration"
    )

    # Create time-based features
    df["day_of_week"] = df["started_at"].dt.day_name()
    df["hour"] = df["started_at"].dt.hour
    df["month"] = df["started_at"].dt.month
    df["month_name"] = df["started_at"].dt.month_name()

    return df


# Apply feature engineering
rides_df = create_features(rides_df)
```

### Ride Duration Validation

Ride duration was calculated in minutes and used as a primary behavioral metric. Records with zero or negative durations were removed, as these indicate invalid or incomplete trips.

Extremely long ride durations were retained at this stage to preserve legitimate edge cases (e.g., extended leisure rides or delayed dockings). The impact of these values is examined during exploratory analysis using distribution-based methods rather than arbitrary filtering.

I engineered duration and temporal features directly from timestamps, removed invalid rides to prevent skewed aggregates, and retained both numeric and human-readable time features to support accurate sorting and clear visualization.

```python
# 🔍 Feature Validation and Sanity Checks

required_cols = [
    "ride_length_min",
    "day_of_week",
    "hour",
    "month"
]

# Verify presence of required engineered features
missing_cols = [col for col in required_cols if col not in rides_df.columns]
if missing_cols:
    print("❌ Missing required columns:", missing_cols)
else:
    print("✅ All required engineered features are present")

# Validate ride duration distribution
print("\nRide length (minutes) summary statistics:")
print(f"  Min: {rides_df['ride_length_min'].min():.2f}")
print(f"  Median: {rides_df['ride_length_min'].median():.2f}")
print(f"  95th percentile: {rides_df['ride_length_min'].quantile(0.95):.2f}")
print(f"  Max: {rides_df['ride_length_min'].max():.2f}")

# Note: Extreme values are intentionally retained for exploratory analysis
# and will be evaluated using distribution-based methods in the Analyze phase.

# Preview engineered features
rides_df[required_cols].head()
```

## Bike Type Usage by Rider Type

This analysis examines whether **casual riders and annual members differ in their usage of available bike types** (classic, electric, and docked) to further contextualize rider behavior.

Comparing bike type preferences across rider segments provides additional descriptive insight into how each group engages with the Cyclistic system. While bike type selection alone does not determine rider intent, consistent differences may reflect broader usage patterns, such as routine transportation versus more flexible or occasional riding.

This analysis serves as a **supporting insight** that complements the primary findings on ride duration and time-based usage. Observed differences are interpreted descriptively and used to enrich the overall behavioral narrative rather than as standalone drivers of marketing strategy.

```python
# Bike type usage distribution by rider type (percentage within rider group)

bike_type_dist = (
    rides_df
    .groupby(["member_casual", "rideable_type"])
    .size()
    .reset_index(name="count")
)

# Calculate percentage distribution within each rider type
bike_type_dist["percentage"] = (
    bike_type_dist
    .groupby("member_casual")["count"]
    .transform(lambda x: x / x.sum() * 100)
)

# Use observed bike types to avoid silent category drops
bike_order = sorted(rides_df["rideable_type"].dropna().unique())

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=bike_type_dist,
    x="rideable_type",
    y="percentage",
    hue="member_casual",
    order=bike_order
)

plt.title("Distribution of Bike Types by Rider Type")
plt.xlabel("Bike Type")
plt.ylabel("Percentage of Rides")
plt.legend(title="Rider Type")

# Format y-axis as percentage
ax.yaxis.set_major_formatter(lambda x, _: f"{x:.0f}%")

plt.tight_layout()
plt.show()
```

### Distribution of Rideable Type

This analysis compares the distribution of bike types used by casual riders and annual members to provide additional context for rider behavior.

**Key Observations:**
- **Electric and classic bikes** account for the majority of rides for both rider types, with broadly similar proportional usage.
- **Docked bikes** are used almost exclusively by casual riders, with negligible usage among annual members.

**Interpretation:**  
The similarity in electric and classic bike usage suggests that bike type preference alone does not strongly differentiate casual riders from annual members. The near-exclusive use of docked bikes by casual riders likely reflects structural or product-related factors—such as pricing models or access rules—rather than differences in riding behavior.

Overall, bike type functions as a **supporting contextual insight**, while more meaningful behavioral differentiation emerges from ride duration and time-based usage patterns examined in subsequent sections.

This capping was applied **only for visualization purposes**. All ride records were retained for analysis, ensuring that summary statistics and behavioral comparisons remain unbiased.

```python
# Ensure day_of_week has correct chronological order
day_order = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
rides_df["day_of_week"] = pd.Categorical(
    rides_df["day_of_week"],
    categories=day_order,
    ordered=True
)

# Aggregate ride counts by day of week and rider type
dow_dist = (
    rides_df
    .groupby(["member_casual", "day_of_week"])
    .size()
    .reset_index(name="count")
)

# Convert counts to percentages within each rider type
dow_dist["percentage"] = (
    dow_dist
    .groupby("member_casual")["count"]
    .transform(lambda x: x / x.sum() * 100)
)

# Plot normalized weekly distribution
plt.figure(figsize=(12, 6))

ax = sns.barplot(
    data=dow_dist,
    x="day_of_week",
    y="percentage",
    hue="member_casual",
    hue_order=["casual", "member"]
)

plt.title("Weekly Ride Distribution by Rider Type")
plt.xlabel("Day of Week")
plt.ylabel("Percentage of Rides")
plt.legend(title="Rider Type")

# Format y-axis as percentage
ax.yaxis.set_major_formatter(lambda x, _: f"{x:.0f}%")

plt.tight_layout()
plt.show()
```
